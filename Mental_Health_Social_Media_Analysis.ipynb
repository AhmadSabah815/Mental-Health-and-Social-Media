## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

print('Libraries imported successfully!')
```

```python
# Data types
mydata.dtypes
```

```python
# Drop rows with missing values (if any)
mydata = mydata.dropna()
print('Shape after dropping nulls:', mydata.shape)
```

## 4. Numeric Columns — Histograms

```python
# Histograms of all numeric columns
mydata.hist(figsize=(14, 10), bins=15, color='steelblue', edgecolor='black')
plt.suptitle('Distribution of Numeric Features', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

## 5. Gender Distribution

```python
# Value counts
mydata['Gender'].value_counts()
```

```python
# Bar chart of gender
mydata['Gender'].value_counts().plot(kind='bar', color=['steelblue','green','orange'], edgecolor='black')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

## 6. Age Analysis

```python
# Age mean
mydata['Age'].mean()
```

```python
# Age standard deviation
mydata['Age'].std()
```

```python
# Age histogram by Gender
mydata.hist(by='Gender', column='Age', bins=10, figsize=(12, 5), color='steelblue', edgecolor='black')
plt.suptitle('Age Distribution by Gender', fontsize=13)
plt.tight_layout()
plt.show()
```

## 7. Social Media Platform Analysis

```python
# Platform value counts
mydata['Social_Media_Platform'].value_counts()
```

```python
# Crosstab: Platform vs Gender
pd.crosstab(mydata['Social_Media_Platform'], mydata['Gender'])
```

```python
# Pivot table: Platform x Gender count
pd.pivot_table(mydata, index=['Social_Media_Platform', 'Gender'], aggfunc='size')
```

```python
# Count plot: Platform by Gender
plt.figure(figsize=(12, 6))
sns.countplot(x='Social_Media_Platform', hue='Gender', data=mydata,
              palette={'Male': 'steelblue', 'Female': 'green', 'Other': 'orange'})
plt.title('Social Media Platform Usage by Gender')
plt.xlabel('Social_Media_Platform')
plt.ylabel('count')
plt.legend(title='Gender')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Stacked bar: Platform by Gender
platform_gender = pd.crosstab(mydata['Social_Media_Platform'], mydata['Gender'], normalize='index') * 100
platform_gender.plot(kind='bar', stacked=True, figsize=(12, 6),
                     color=['green', 'orange', 'steelblue'], edgecolor='black')
plt.title('Platform Usage by Gender (% Stacked)')
plt.xlabel('Social_Media_Platform')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=15)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

## 8. Daily Screen Time Analysis

```python
# Screen time stats
print('Mean:', mydata['Daily_Screen_Time(hrs)'].mean())
print('Median:', mydata['Daily_Screen_Time(hrs)'].median())
print('Std:', mydata['Daily_Screen_Time(hrs)'].std())
```

```python
# Avg screen time by platform
mydata.groupby('Social_Media_Platform')['Daily_Screen_Time(hrs)'].mean().sort_values(ascending=False).plot(
    kind='bar', color='steelblue', edgecolor='black', figsize=(10, 5))
plt.title('Average Daily Screen Time by Platform')
plt.xlabel('Platform')
plt.ylabel('Avg Hours/Day')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Boxplot: Screen time by gender
plt.figure(figsize=(8, 5))
sns.boxplot(x='Gender', y='Daily_Screen_Time(hrs)', data=mydata,
            palette={'Male': 'steelblue', 'Female': 'green', 'Other': 'orange'})
plt.title('Screen Time by Gender')
plt.tight_layout()
plt.show()
```

## 9. Sleep Quality & Stress Level

```python
# Avg sleep quality by platform
mydata.groupby('Social_Media_Platform')['Sleep_Quality(1-10)'].mean().sort_values(ascending=False).plot(
    kind='bar', color='teal', edgecolor='black', figsize=(10, 5))
plt.title('Average Sleep Quality by Platform')
plt.xlabel('Platform')
plt.ylabel('Avg Sleep Quality (1-10)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Avg stress level by platform
mydata.groupby('Social_Media_Platform')['Stress_Level(1-10)'].mean().sort_values(ascending=False).plot(
    kind='bar', color='salmon', edgecolor='black', figsize=(10, 5))
plt.title('Average Stress Level by Platform')
plt.xlabel('Platform')
plt.ylabel('Avg Stress Level (1-10)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 10. Happiness Index Analysis

```python
# Avg happiness by platform
mydata.groupby('Social_Media_Platform')['Happiness_Index(1-10)'].mean().sort_values(ascending=False).plot(
    kind='bar', color='gold', edgecolor='black', figsize=(10, 5))
plt.title('Average Happiness Index by Platform')
plt.xlabel('Platform')
plt.ylabel('Avg Happiness (1-10)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Scatter: Screen time vs Happiness
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Daily_Screen_Time(hrs)', y='Happiness_Index(1-10)',
                hue='Gender', data=mydata,
                palette={'Male': 'steelblue', 'Female': 'green', 'Other': 'orange'})
plt.title('Screen Time vs Happiness Index')
plt.tight_layout()
plt.show()
```

## 11. Correlation Heatmap

```python
# Select numeric columns
numeric_cols = mydata.select_dtypes(include=np.number).drop(columns=['User_ID'], errors='ignore')

plt.figure(figsize=(10, 7))
sns.heatmap(numeric_cols.corr(), annot=True, fmt='.2f', cmap='coolwarm',
            linewidths=0.5, square=True)
plt.title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.show()
```

## 12. Exercise Frequency & Days Without Social Media

```python
# Exercise frequency distribution
mydata['Exercise_Frequency(week)'].value_counts().sort_index().plot(
    kind='bar', color='mediumseagreen', edgecolor='black', figsize=(8, 5))
plt.title('Exercise Frequency per Week')
plt.xlabel('Days per Week')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Days without social media distribution
mydata['Days_Without_Social_Media'].value_counts().sort_index().plot(
    kind='bar', color='mediumpurple', edgecolor='black', figsize=(8, 5))
plt.title('Days Without Social Media')
plt.xlabel('Days')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 13. Summary Statistics by Platform

```python
# Full summary grouped by platform
mydata.groupby('Social_Media_Platform')[[
    'Daily_Screen_Time(hrs)',
    'Sleep_Quality(1-10)',
    'Stress_Level(1-10)',
    'Happiness_Index(1-10)',
    'Exercise_Frequency(week)'
]].mean().round(2)
```
